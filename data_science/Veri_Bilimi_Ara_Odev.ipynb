```python
def normalize_et(series):

  return(
      series.astype(str)
      .str.strip()
      .str.lower()
      .str.replace("i̇", "i", regex=False)
      .str.replace("ı", "i", regex=False)
      .str.replace("ö", "o", regex=False)
      .str.replace("ü", "u", regex=False)
      .str.replace("ç", "c", regex=False)
      .str.replace("ş", "s", regex=False)
      .str.replace("ğ", "g", regex=False)
  )
```

# 11. kategori sütunundaki tutarsız değerleri inceleyin. Örneğin Ev Yaşam, ev yaşam, ev&yaşam gibi farklı yazımları tek biçime dönüştürün. Temizlikten sonra value_counts() çıktısını yazdırın.

```python
df["kategori"].unique()
```

```python
def normalize_et(series):

  return(
      series.astype(str)
      .str.strip()
      .str.lower()
      .str.replace("i̇", "i", regex=False)
      .str.replace("ı", "i", regex=False)
      .str.replace("ö", "o", regex=False)
      .str.replace("ü", "u", regex=False)
      .str.replace("ç", "c", regex=False)
      .str.replace("ş", "s", regex=False)
      .str.replace("ğ", "g", regex=False)
  )
```

```python
kategori_duzeltme = {
    "elektronik": "Elektronik",
    "kitap": "Kitap",
    "kozmetik": "Kozmetik",
    "kozmetık": "Kozmetik",
    "spor": "Spor",
    "ev yasam": "Ev Yaşam",
    "ev ve yasam": "Ev Yaşam",
    "ev&yasam": "Ev Yaşam",
    "ofis": "Ofis"
}
```

```python
df["kategori"] = normalize_et(df["kategori"])
```

```python
df["kategori"] = df["kategori"].replace(kategori_duzeltme)
```

```python
df["kategori"].value_counts()
```

```python
# önce boşluklardan  ı i gibi yanlış sesli harf tercihlerinden kurtulduk
# daha sonra düzeltme sözlüğü oluşturarak son düzeltmeleri yaptık
```

# 13. birim_fiyat sütunu için IQR yöntemini kullanarak Q1, Q3, IQR, alt sınır ve üst sınır değerlerini hesaplayın. Sonra aykırı değer içeren kayıt sayısını yazdırın.

```python
q1 = df["birim_fiyat"].quantile(0.25)
q3 = df["birim_fiyat"].quantile(0.75)
```

```python
iqr = q3 - q1
alt_sinir = q1 - 1.5 * iqr
ust_sinir = q3 + 1.5 * iqr
```

```python
len(df[(df["birim_fiyat"] < alt_sinir) | (df["birim_fiyat"] > ust_sinir)])
```

```python
#quantile fonksiyonunu kullanarak q1 ve q3 ü bulduk k değerini 1.5 aldık
#filtreleme yapıp hatalı satırların sayısını bulduk
```

# 14. toplam_tutar sütunu için histogram çizdirin ve box plot çizdirin. Ardından dağılım hakkında 2–3 cümlelik kısa bir yorum yazın.

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(12, 5))

# Histogram
plt.subplot(1, 2, 1)
sns.histplot(x=df["toplam_tutar"], bins=30, kde=True, color="blue")
plt.title("Toplam Tutar - Histogram")
plt.xlabel("Toplam Tutar")
plt.ylabel("Kayıt Sayısı")

# Box Plot
plt.subplot(1, 2, 2)
sns.boxplot(x=df["toplam_tutar"], color="orange")
plt.title("Toplam Tutar - Box Plot")
plt.xlabel("Toplam Tutar")

plt.tight_layout()
plt.show()
```
