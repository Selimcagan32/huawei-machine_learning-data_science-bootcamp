# Veri Bilimi Eğitimi


## Veri Setini Tanıma

Amaç:
  1. Veri setinin genel yapısını tanımak ve temel özelliklerini incelemek
  2. Veri analizi öncesi gerekli ilk kontrolleri uygulamak

Adımlar:
  1. Veri setini yükle
  2. Veri setinin ilk ve son satırlarını incele
  3. Veri setinin boyutunu ve sütunlarını kontrol etmek
  4. Veri setinin genel bilgisini incele
  5. Özet istatistikleri görüntüle
  6. Veri tiplerini kontrol et
  7. Sayısal ve kategorik sütunları ayırmak
  8. Benzersiz (unique) değer sayılarını incelemek

```python
import pandas as pd
```

```python
# 1. Veri setini yükleme
df = pd.read_csv("e_ticaret_veri_seti.csv")
```

```python
# 2. Veri setinin ilk satırlarını inceliyoruz
df.head()
```

```python
# 2. Veri setinin son satırlarını inceliyoruz
df.tail()
```

```python
# 3. Veri setinin boyutunu ve sütunlarını kontrol etme
veri_seti_boyut = df.shape
veri_seti_boyut
```

```python
# satırlarını kontrol etme
satir_sayisi = df.shape[0]
print(f"satır sayısı: {satir_sayisi}")
```

```python
# sütunlarını kontrol etme
sutun_sayisi = df.shape[1]
print(f"sütun sayısı: {sutun_sayisi}")
```

```python
# sütun isimleri
for i, sutun in enumerate(df.columns, start = 1): # enumerate döngüye sokulan elemanları saymak için kullanılan bir fonksiyondur.
  print(f"{i}. {sutun}") # start = 1 sayma işleminin 1'den başlaması için
```

```python
# 4. Veri setinin genel bilgisini inceleme
df.info() # sütun isimlerini, veri tiplerini, kayıp, eksik verileri görebiliriz
```

```python
# 5. Özet istatistikleri görüntüleme
df.describe() # sayısal sütunlar için temel istatistikleri yazdırır
```

```python
# 6. Veri tiplerini kontrol etme
df.dtypes #  df.info() komutuna benzer şekilde sütunların hangi türde veriler içerdiğini gösterir
```

```python
# 7. Sayısal ve kategorik sütunların ayrılması
# sayısal: float64, int64
sayisal_sutunlar = df.select_dtypes(include=["int64", "float64"]).columns
sayisal_sutunlar
```

```python
# kategorik: object
kategorik_sutunlar = df.select_dtypes(include=["object"]).columns
kategorik_sutunlar
```

```python
# 8. Benzersiz (unique) değer sayılarını incelemek
for sutun in df.columns:
  print(f"{sutun}: {df[sutun].nunique()}")
```
